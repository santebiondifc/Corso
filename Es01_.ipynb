Importare Pandas e matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt

escola = pd.read_csv("matriculas_escola.csv")
print ("Linee e colonne",escola.shape)
escola.head()
```

Pulizia del file con .isna().mean()*100).round(1)

```python
mancanti = (escola.isna().mean()*100).round(1)
print(mancanti.sort_values(ascending=False))
```

Pulizia "curso"

```python
print(escola["curso"].value_counts())
```

puliamo tutte le colonne

```python
escola["aluno"] = escola["aluno"].str.strip().str.upper()
escola["responsavel"] = escola["responsavel"].str.strip().str.upper()
escola["curso"] = escola["curso"].str.strip().str.upper()
escola["turno"] = escola["turno"].str.strip().str.upper()
escola["situacao"] = escola["situacao"].str.strip().str.upper()
escola["situacao"] = escola["situacao"].str.strip().str.upper()
print(escola["curso"].value_counts())
```

Togliere i doppioni

```python
de_para_curso={
    "TÉCNICO EM ADMINISTRAÇÃO":"Tecnico em aministracao",
    "TÉCNICO EM INFORMÁTICA":"Tecnico em informatica",
    "TÉCNICO EM SEGURANÇA DO TRABALHO":"Tecnico em seguranca do trabalho",
    "TECNICO EM ADMINISTRACAO":"Tecnico em aministracao",
    "TÉCNICO EM ENFERMAGEM":"Tecnico em enfermagem",
    "TÉCNICO EM ESTÉTICA":"Tecnico em estetica",
    "TÉCNICO EM LOGÍSTICA":"Tecnico em logistica",
    "TÉCNICO EM CONTABILIDADE":"Tecnico em contabilidade",
    "ENFERMAGEM":"Tecnico em enfermagem",
    "TECNICO EM INFORMATICA":"Tecnico em informatica",
    "SEGURANCA DO TRABALHO":"Tecnico em seguranca do trabalho",
    "TECNICO EM LOGISTICA":"Tecnico em logistica",
    "TECNICO EM CONTABILIDADE":"Tecnico em contabilidade",
    "TECNICO EM ESTETICA":"Tecnico em estetica",
    "TECNICO EM ENFERMAGEM":"Tecnico em enfermagem"
}
escola["curso"] = escola["curso"].replace(de_para_curso)
print(escola["curso"].value_counts())
```

```python
print(escola["turno"].value_counts())
```

turno

```python
de_para_turno = {
    "NOITE":"Noite",
    "MANHÃ":"Manha",
    "TARDE":"Tarde"
}
escola["turno"] = escola["turno"].replace(de_para_turno)
print(escola["turno"].value_counts())
```

situacao


```python
print(escola["situacao"].value_counts())
```

```python
de_para_situacao = {
    "CURSANDO":"Cursando",
    "EVADIDO":"Evadido",
    "CONCLUÍDO":"Concluido",
    "CONCLUIDO":"Concluido"
}
escola["situacao"] = escola["situacao"].replace(de_para_situacao)
print(escola["situacao"].value_counts())
```

duplicatas — a mesma pessoa, no mesmo curso, na mesma data: é rematrícula digitada duas vezes.

```python
escola["data_matricula"]=pd.to_datetime(escola["data_matricula"], errors="coerce")
print(escola["data_matricula"].head())
```

```python
duplicada=escola.duplicated(subset=["aluno", "data_matricula","curso"])
nome_completo=escola["aluno"].str.split().str.len()>=2
tem_situacao=escola["situacao"].notna()
print("Base bruta:                ",len(escola))
print("Sem duplicatas:            ",(~duplicada).sum())
print("Com nome completo:         ",(~duplicada & nome_completo).sum())
print("Com situação:              ",(~duplicada & nome_completo & tem_situacao).sum())
```

Salvare il nuovo elaborato

```python
trattato=escola[~duplicada & nome_completo & tem_situacao]
trattato.to_csv("escola_trattato.csv",index=False)
print("Base trattata:", len(trattato),"linee")
```

Le tre domande:
Qual è il tasso di abbandono scolastico? E come varia a seconda del corso di studi?

```python
taxa_evasao = trattato["situacao"].eq("Evadido").mean() * 100

print("Total de alunos:", len(trattato))
print("Alunos evadidos:", trattato["situacao"].eq("Evadido").sum())
print("Taxa de abandono:", round(taxa_evasao, 1), "%")
```

```python
taxa_por_curso = (
    trattato.groupby("curso")["situacao"]
    .apply(lambda x: x.eq("Evadido").mean() * 100)
    .round(1)
)

print(taxa_por_curso)
```

Qual turno concentra a evasão? A escola tem mais aula à noite — isso ajuda ou atrapalha?

```python
evasao_turno = (
    trattato.groupby("turno")["situacao"]
    .apply(lambda x: x.eq("Evadido").mean() * 100)
    .round(1)
)

print(evasao_turno)
```

Quantos alunos distintos a escola atendeu de fato no período?



```python
alunos_distintos = trattato[
    trattato["situacao"].isin(["Cursando", "Concluido"])
]["aluno"].nunique()

print("Alunos distintos atendidos:", alunos_distintos)
```

Responsavel

```python
evadidos = trattato[trattato["situacao"] == "Evadido"]

print("Total evadidos:", len(evadidos))
print("Com responsável:", evadidos["responsavel"].notna().sum())
print("Sem responsável:", evadidos["responsavel"].isna().sum())
taxa_com_responsavel = evadidos["responsavel"].notna().mean() * 100

print(
    "Evadidos com responsável:",
    round(taxa_com_responsavel, 1),
    "%"
)
```

Associato ad ogni responsabile

```python
evadidos_por_responsavel = (
    trattato[trattato["situacao"] == "Evadido"]
    .dropna(subset=["responsavel"])
    .groupby("responsavel")["aluno"]
    .nunique()
    .sort_values(ascending=False)
)

print(evadidos_por_responsavel)
```
